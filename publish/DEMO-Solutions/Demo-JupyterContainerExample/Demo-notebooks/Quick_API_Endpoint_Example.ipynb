## Set Up API Details

## Prepare the Request Payload

We'll send a simple prompt to test if the API responds.

```python
headers = {
 "Authorization": f"Bearer {API_KEY}",
 "Content-Type": "application/json"
}

payload = {
 "model": DEPLOYMENT_NAME,
 "messages": [
 {"role": "system", "content": "You are a helpful assistant."},
 {"role": "user", "content": "What AI model are you?"}
 ]
}
```

## Send the Request and Check Response

If the API is working, you should see a successful response (status code 200) and the generated content.

```python
import requests
import json

response = requests.post(ENDPOINT_URL, headers=headers, data=json.dumps(payload))

if response.status_code == 200:
 # Format the JSON for easier reading
 formatted_response = json.dumps(response.json(), indent=4)
 print("API is working! Raw response:")
 print(formatted_response)
else:
    print(f"Error: {response.status_code}")
    print(response.text)
```

## Print out the Response
This parses the JSON response and prints the generated content from the AI model.

```python
response_data = response.json()
generated_content = response_data['choices'][0]['message']['content']
# Check if generated_content is empty and if it is look for the reasoning_content instead of content field
if generated_content.strip() == "":
    generated_content = response_data['choices'][0]['message']['reasoning_content']
print("Generated Content from AI Model:")
print(generated_content)
```
